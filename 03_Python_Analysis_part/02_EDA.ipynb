```python
# Import pandas library to read the data.
import pandas as pd
import os

# Import data profiling, reporting and operating system libraries.
from data_profiling import ProfileReport

# Import the connection of the database engine.
from _01_database_connection import engine

# Report path.
report_path = r"data_quality_report/"
os.makedirs(report_path, exist_ok=True)
```

```python
# 1) read fact_sales table.
fact_sales = pd.read_sql(
    sql="SELECT * FROM gold.fact_sales",
    con=engine
)

# Pick a sample.
fact_sales.sample(5)
```

```python
# Report about the fact_sales table.
sales_profile = ProfileReport(
    df=fact_sales,
    title="Sales Profile Report",
    explorative=True
)

sales_profile.to_file(f"{report_path}01_fact_sales_report.html")
```

```python
# Inspect datat ypes.
fact_sales.dtypes
```

```python
# Inspect missing values.
fact_sales.isna().sum()
```
